# PER vs Energy figure template

Fill in the manual statistics below, then run the plotting cells to generate a two-axis comparison figure.


```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42
mpl.rcParams['font.family'] = 'sans-serif'
mpl.rcParams['axes.spines.top'] = True

GREY = '#8C8C8C'
MATTE_BLUE = '#2F4F6F'
```

```python
# Manually enter your summary statistics here.
# Use np.nan as a placeholder until values are available.

models = ['LSTM', 'MLP', 'SOTA GRU', 'SNN']

per_values = np.array([
    np.nan,  # LSTM PER
    np.nan,  # MLP PER
    0.108,  # SOTA GRU PER
    0.128,  # SNN PER
], dtype=float)

energy_uJ = np.array([
    np.nan,  # LSTM energy per inference (uJ)
    np.nan,  # MLP energy per inference (uJ)
    18,  # SOTA GRU energy per inference (uJ)
    4,  # SNN energy per inference (uJ)
], dtype=float)

# Optional error bars. Leave as zeros if you do not want error bars.
per_err = np.array([0.0, 0.0, 0.0, 0.0], dtype=float)
energy_err = np.array([0.0, 0.0, 0.0, 0.0], dtype=float)

bar_width = 0.36
figure_size = (6.2, 3.8)
save_path = 'per_energy_comparison.png'
```

```python
x = np.arange(len(models))
width = 0.1

fig, ax1 = plt.subplots(figsize=figure_size)
ax2 = ax1.twinx()

per_bars = ax1.bar(
    x - width / 2,
    per_values,
    width=width,
    color=GREY,
    edgecolor=GREY,
    yerr=per_err,
    capsize=4,
    zorder=3,
)

energy_bars = ax2.bar(
    x + width / 2,
    energy_uJ,
    width=width,
    color=MATTE_BLUE,
    edgecolor=MATTE_BLUE,
    yerr=energy_err,
    capsize=4,
    zorder=3,
)


ax1.set_xticks(x)
ax1.set_xticklabels(models, rotation=15)
ax1.set_ylabel('Phoneme Error Rate (PER)', color=GREY, fontweight='bold')
ax2.set_ylabel('Energy Per Inference (uJ)', color=MATTE_BLUE, fontweight='bold')

ax1.tick_params(axis='y', colors=GREY)
ax2.tick_params(axis='y', colors=MATTE_BLUE)
ax1.tick_params(axis='x', length=0)

ax1.set_axisbelow(True)

ax1.spines['left'].set_color(GREY)
ax2.spines['right'].set_color(MATTE_BLUE)
ax1.spines['bottom'].set_color('#4A4A4A')
ax1.spines['top'].set_color('#4A4A4A')

ax1.margins(x=0.08)
fig.tight_layout()
plt.show()
```

```python
# Optional: save after you have filled in the values above.
x = np.arange(len(models))

fig, ax1 = plt.subplots(figsize=figure_size)
ax2 = ax1.twinx()

ax1.bar(x - bar_width / 2, per_values, width=bar_width, color=GREY, edgecolor=GREY, yerr=per_err, capsize=4)
ax2.bar(x + bar_width / 2, energy_uJ, width=bar_width, color=MATTE_BLUE, edgecolor=MATTE_BLUE, yerr=energy_err, capsize=4)

ax1.set_xticks(x)
ax1.set_xticklabels(models, rotation=15)
ax1.set_ylabel('Phoneme Error Rate (PER)', color=GREY, fontweight='bold')
ax2.set_ylabel('Energy Per Inference (uJ)', color=MATTE_BLUE, fontweight='bold')
ax1.tick_params(axis='y', colors=GREY)
ax2.tick_params(axis='y', colors=MATTE_BLUE)
ax1.tick_params(axis='x', length=0)
ax1.grid(axis='y', alpha=0.22, linewidth=0.8)
ax1.set_axisbelow(True)
ax1.spines['left'].set_color(GREY)
ax2.spines['right'].set_color(MATTE_BLUE)
ax1.spines['bottom'].set_color('#4A4A4A')
ax1.spines['top'].set_color('#4A4A4A')
ax1.margins(x=0.08)

fig.tight_layout()
fig.savefig(save_path, dpi=300, bbox_inches='tight')
plt.show()

print(f'Saved figure to: {save_path}')
```
